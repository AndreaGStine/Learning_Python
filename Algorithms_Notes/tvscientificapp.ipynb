```python
# Q1) Define a function that returns the frequencies of the last digits 
#     of a list of nonnegative integers. 
#
#     Use only the Python standard libraries.
#
#     Given the list [49, 10, 20, 5, 30, 785]: 
#     9 is the last digit once (in 49), 
#     0 is the last digit three times (in 10, 20, and 30), 
#     5 is the last digit two times (in 5 and 785) 
# last_digit_counts([49, 10, 20, 5, 30, 785]) 
#   = {9:1, 0:3, 5:2} # or something equivalent

from collections import defaultdict
def last_digit_counts(nums):
    counts = defaultdict(int)
    for num in nums:
        counts[num % 10] += 1
    return dict(counts)
```

```python
print(last_digit_counts([49, 10, 20, 5, 30, 785]))
```

```python
# Q2) Given a list of 2-D coordinates, write a function that returns True
#     if the points lie on a straight line and False otherwise.
#
#     Use only the Python standard libraries.
#
# collinear([[1,1], [2,2], [4,4], [-10, -10]]) = True
# collinear([[1,0], [2,0], [3,1]]) = False

def collinear(points):
    if len(points) < 2:
        return True
    
    xchange = 0
    ychange = 0
    differentpoints = False
    for i in range(1, len(points)):
        if points[i] != points[0] and not differentpoints:
            ychange = points[i][1] - points[0][1]
            xchange = points[i][0] - points[0][0]
            differentpoints = True
        else: 
            if xchange != 0:
                if (points[i][0] - points[i-1][0]) !=  0:
                    if (points[i][1] - points[i-1][1]) / (points[i][0] - points[i-1][0]) != ychange/xchange:
                        return False
                else:
                    return False
            else:
                if points[i][0] - points[i-1][0] != 0:
                    return False
    return True
```

```python
print(collinear([[1,1], [2,2], [4,4], [-10, -10]]))
print(collinear([[1,0], [2,0], [3,1]]))
print(collinear([[0,1],[1,1]]))
print(collinear([[0,1],[0,2]]))
print(collinear([[0,0],[0,0],[1,1],[3,2]]))
print(collinear([[0,0],[0,0],[1,1],[3,3]]))
```
